```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('WideTransformationsInRDD').getOrCreate()
sc = spark.sparkContext
```

```python
# groupByKey()

rdd = sc.parallelize([('a', 1), ('b', 2), ('a', 3)])

result = rdd.groupByKey()
```

```python
result.collect()
```

```python
result.mapValues(list).collect()
```

```python
# reduceByKey(func)

result = rdd.reduceByKey(lambda x,y:x+y)
result.collect()
```

```python
# aggregateByKey(zeroValue, seqFunc, combFunc)

result = rdd.aggregateByKey(0, lambda x,y:x+y, lambda acc1,acc2:acc1+acc2)
result.collect()
```

```python
# combineByKey(createCombiner, mergeValue, mergeCombiners)

result = rdd.combineByKey(
    lambda v:[v],
    lambda comb,v:comb+v,
    lambda c1,c2: c1+c2
)
result.collect()
```

```python
# sortByKey([ascending], [numPartitions])

rdd = sc.parallelize([(3,'a'),(1,'b'),(2,'c')])
result = rdd.sortByKey(ascending=True)
result.collect()
```

```python
result = rdd.sortByKey(ascending=False)
result.collect()
```

```python
# sortBy(func, ascending)

rdd = sc.parallelize(['hello','cat','highest','hi'],2)
result = rdd.sortBy(lambda x:len(x), ascending=True)
result.collect()
```

```python
# Inner join - join(otherRDD)
rdd1 = sc.parallelize([('a',1),('b',2)])
rdd2 = sc.parallelize([('c',3),('b',4)])
result = rdd1.join(rdd2)
result.collect()
```

```python
# Left outer join, Right outer join, fullOuter join
result = rdd1.leftOuterJoin(rdd2).collect()
result = rdd1.rightOuterJoin(rdd2).collect()
```
