## RDD (Resilient Distributed Datasets)

### RDDs in PySpark are fundamental, fault-tolerant distributed collection of objects that can be operated in parallel across a cluster.

### RDDs are immutable but new RDDs can be created from existing ones through transformation operations.

### RDDs are designed for fault tolerance with automatic recovery, allowing efficient distributed computing on big data.

### What is an RDD?

#### Resilient: Automatically recovers from failures.

#### Distributed: Data is split across multiple nodes in the cluster.

#### Dataset: Collection of elements/objects.

#### Immutable: Cannot be changed once created; transformations create new RDDs.

#### Supports parallel operations like map, filter, reduce across the data.

### RDD Operations: Transformations and Actions

#### Transformations (lazy evaluation)

#### Transformations create a new RDD from an existing one. 
#### They are lazy, meaning they do not compute immediately but build a lineage graph.

#### Common transformations:

#### map(func): Applies a function to each element.

#### filter(func): Filters elements based on a condition.

#### flatMap(func): Maps and flattens results.

#### union(otherRDD): Combines two RDDs.

#### join(otherRDD): Joins two RDDs by their keys.

#### reduceByKey(func): Aggregates values by key.

### Actions (trigger execution)
#### Actions trigger the execution of transformations and return results:

#### Common actions:

#### collect(): Returns all elements of the RDD as an array.

#### count(): Returns the number of elements.

#### reduce(func): Reduces elements using a specified function.

#### take(n): Returns the first n elements.

#### saveAsTextFile(path): Saves the dataset to a text file.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('SparkExample').getOrCreate()
```

```python
# Creating an empty RDD
emptyRDD = spark.sparkContext.emptyRDD()
print(emptyRDD)
```

```python
# Creating an empty RDD using parallelize() method
emptyRDD2 = spark.sparkContext.parallelize([])
print(emptyRDD2)
```

```python
# Define spark context variable
sc = spark.sparkContext
```

```python
words = sc.parallelize(["scala","Python","Java","hadoop","snowflake","kafka","scala","Java"])
```

```python
words.count()
```

```python
words.collect()
```

```python
# Map Transformation
# Lazy evaluation
words_length = words.map(lambda x : (x,len(x)))
```

```python
# Lazy evaluation - Above transformation will be applied after below action function is invoked
words_length.collect()
```

```python
# Filter Transformation
long_words = words.filter(lambda x: len(x)>5)
long_words.collect()
```

```python
# Reduce action or Aggregation
numbers = sc.parallelize([1,2,3,4,5])
total = numbers.reduce(lambda a,b : a+b)
total
```

```python
lines = sc.textFile("/Users/SWAROOP/Documents/DATA ENGINEERING/Projects/pyspark-playground/Basics/data/sample.txt")
```

```python
words = lines.flatMap(lambda line: line.split(" "))
```

```python
word_tuples = words.map(lambda word: (word,1))
```

```python
word_counts = word_tuples.reduceByKey(lambda a,b: a+b)
```

```python
print(word_counts.collect())
```
